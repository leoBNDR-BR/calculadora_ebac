```python
num1 = input("Digite o número 1: ")
num2 = input("Digite o número 2: ")

num_conv1 = int(num1)
num_conv2 = int(num2)

##print ('num_conv1', type(num_conv1))
##print ('num_conv2', type(num_conv2))

adi = num_conv1 + num_conv2 ##print ('adição', adi)
sub = num_conv1 - num_conv2 ##print ('subtração', sub)
mult = num_conv1 * num_conv2 ##print ('multiplicação', mult)
div = num_conv1 / num_conv2 ##print ('divisão', div)

print ('\nEscolha a operação matématica:'
      '\n1 - Soma'
      '\n2 - Subtração'
      '\n3 - Multiplicação'
      '\n4 - Divisão'
      '\n\n5 - Sair!')

opcao = input('\nDigite a opção desejada: ')

if opcao == '1':
    print ('\nSoma: ', adi)
elif opcao == '2':
    print ('\nSubtração: ', sub)
elif opcao == '3':
    print ('\nMultiplicação: ', mult)
elif opcao == '4':
    print ('\nDivisão: ', div)
elif opcao == '5':
    print ('\nSaindo...')
else:
    print ('\nOpção inválida!')

```
